# ML Breadth

## 1. Core Supervised Learning Fundamentals

### 1.1 Problem framing
1. classification vs. regression.\
Given a set of input. Classification problems try to classify response variables into different groups (**discrete classes**). Regression problems try to predict a (**continuous**) value of the response variable.

2. Ranking vs. Classification \
Classification predicts one or more labels for each item independently. Ranking predicts a relative ordering of items, usually to optimize which items should appear higher for a given query, user, or context.

3. multi-class vs. milti-label \
Multi-class means there are multiple classes that separates the inputs; multi-label means each input might get multiple labels.

4. Binary classification vs. anomaly detection \
Binary classification is classifying inputs into two categories. Anomaly detection is you treat the input as if they are from a single distribution, and you try to identify outliers from your dataset.

5. Pointwise, pairwise, and listwise setups \
This usually comes up in ranking/recommendation problems. Pointwise method scores each item independently, pairwise methods learn preferences between item pairs, and listwise methods optimize over the full ranked list. As you move from pointwise to listwise, you generally model ranking structure more directly, but training becomes more complex.

6. When the prediction unit is wrong \
The prediction unit is wrong when a model makes predictions at a different granularity than the real decision or business action. For example, if the intervention happens at the user level but the model predicts at the event level, then the labels, features, and metrics may all become misaligned. This can lead to a model that performs well offline but is not useful in production.

7. Label leakage \
**Different from train-test contamination**
Label leakages occurs when features contain information that directly or indirectly reveals the target, especially information that would not be available at the time of prediction. This makes model performance look unrealistically good. (e.g. a features that's simply response * 2)

8. Proxy labels vs. True Objective
A true objective is the real outcome we ultimately care about, while a proxy label is a measurable substitute used because the true objective is hard, delayed, or expensive to obtain. For example, if I develop an app and want to know how popular it is, I might want to use number of users, avg time a user spends on the app, etc. to infer the popularity. The main risk is that optimizing the proxy may not fully optimize the true goal.

9. Offline metric vs. Business metric mismatch
The offline metric measures model performance on historical or benchmark data, while business metrics measure actual product or oranizational outcomes in the real world. e.g. An offline metric might be whether a chatbot answers all questions correctly. Business metric, on the other hand, might care about the bot's popularity and usage. So sometimes a good offline metric does not automatically guarantee business success.

### 1.2 Bias Variance and Generalization
1. bias vs variance \
Bias is the systematic error from overly strong (and simple) assumptions on the data generating process. Because of those assumptions, the model may be too simple to capture all the patterns in the true data. Variance is the error from the model learning too well on specific examples, making the model not generalize well to unseen data.

2. underfitting vs overfitting \
A model is underfitting if it does not do well even on the training data - it hasn't learned all the generic patterns displayed in the training data. Overfitting happens when a model fits a training data too well, that it even fits the noise in the data, so it does not generalize well to unseen data.

3. train error vs validation error patterns
**This question is about diagnostics** \
If high training error & high val error --> Underfitting  \
If low training error & high val error --> Overfitting \
If low training error & low val error --> Good training \

4. variance reduction methods \
a. Adding more training data. \
b. Reduce parameters to the data. \
c. Stronger regularization. \
d. Stop early \
e. Bagging \

5. regularization intuition \
Regularization is a technique that discourages (e.g. adds penalty terms, constraints, or early stopping) the model from becoming too complex. Essentially what it does is it tries to reduce model complexity without hurting too much on model performance (on the training data). So it does not overfit to the training data.

6. data size effects \
More representative and diverse data helps improve generalization. However, training with more data that is not adding additional information about the population (e.g. duplicate data) is going to give model false confidence about a very narrow pattern in the data.

7. model complexity tradeoffs \
More complex models can capture reacher patterns, but they are harder to train and more prone to overfitting. Simpler models are more interpretable and easier to training, but they might underfit the data and have a high bias.

8. why ensembles help \
* Bagging combines results from unstable learners and reduces variane (intuition is weak learners make different mistakes, and aggregating different weak learners help cancels out noise).
* Boosting reduces bias by sequentially corrects mistakes being made.
* Stacking trains a meta-adaptor that combines the results of multiple models. It helps when the models have complimentary strengths and make different kinds of mistakes.

### 1.3 Train / validation / test

1. purpose of each split \
* The training set is for training the parameters of the model.
* The validation set is for
  * Model selection.
  * Choose hyper-parameters.
  * Choose decision thresholds.
  * Decide when to stop training.
* The test set is never touched during training and only used to evaluate model performance after the training.

2. cross-validation \
Cross-validation is a hyper-parameter tuning technique. The idea is to split the data into K folds, each time using 1 new fold as the validation set and the other K-1 folds as the training set. Then take the model with best average performance.

3. stratified split \
It's a split method that first puts data into classes, and then in each split, maintains the roughly same proportion of each class in the split. E.g. if 10% of the full dataset has positive labels, then in train/val/test split they should also each have 10% positive data.

4. time-based split \
A time-based split is used when data has a temporal order. In this case, should train on earlier-time data and validate/test on later periods. This avoids leaking future information into training.

5. group split \
Sometimes you have data that is in different groups. E.g. you have multiple medical records from the same patient. You want them to go into one split, as they might be highly-duplicated to appear as leakage in data.

6. leakage through preprocessing \
Leakage through preprocessing happens when you compute transformations/pre processing statistics using information from val/test data. E.g.
* Scaling: Should scale data after the split, not scale then split.
* Imputation: Should impute missing data after the split.
* Feature seletion: Should calculate correlation of the features after the split.

7. data contamination
When some data appears both in the training set and test/val set, creating a false increase in the test/val performance.

8. early stopping \
It is a regularization method used to stop the model from learning too much detail about the dataset and prevent overfitting. Usually applied when performance on the val dataset stops improving.

9. hyperparameter tuning without test-set leakage
Whole workflow:
  * Split the data into train/val/test.
  * Train models with different hyper-parameters to train.
  * Compare them on the validation set.
  * Choose the best hyper-parameters.
  * Test once on the test set.

## 2. Linear Models

### 2.1 Linear regression
1. what it optimizes \
Linear regression optimizes the residual sum of squares between the predicted value and the true value given a set of inputs (features).

2. mean squared error \
It is the loss function for linear regression training. The average of the squred prediction error. It penalizes large errors more than small ones, and it is always positive.

3. ordinary least squares \
The closed-form OLS solution sets the gradient of SSR to zero:

$$\nabla_{\boldsymbol{\beta}}\,\text{SSR} = -2\mathbf{X}^\top(\mathbf{y} - \mathbf{X}\boldsymbol{\beta}) = 0$$

Solving gives the **normal equations**:

$$\hat{\boldsymbol{\beta}} = (\mathbf{X}^\top \mathbf{X})^{-1} \mathbf{X}^\top \mathbf{y}$$

under Gauss-Markov assumptions (below), OLS is **BLUE** - the best linear unbiased estimator.

4. assumptions: \
    * linearity: The conditional mean of y given x is linear. If this breaks, then the predictions are likely biased.

    * independent errors: The errors (noise) are uncorrelated. If this breaks, the linear regression is still unbiased, but the significance on the coefficients can be over-estimated (over-confident). Intuitively speaking, the correlation in residual will reduce the amount of independent information in the data, and in tern increase the variance (and also sd) of beta and boosts tests like t-statitsics. Another useful mental model is that if you have 100 data and they are 5 groups of highly-correlated data, then you are effectively only having 5 'informative' data, but it 'looks like' you have a large sample./

    * homoscedasticity: It means that the variance of the errors is constant around different values of x. In simpler term, the 'thickness' or 'width' of y values at different x values are roughly the same (**NOT** the number of y points associated with each x, just the thickness). If this is violated, then again the variance of beta hat does not follow the normal formula, and becomes unpredicted, and it's hard to tell whether the statistics around beta hat is still accurate or not.

    * low multicollinearity: Meaning that the no features (columns in the X matrix) is a linear combination of the others. If this is violated, then the solution is not stable (small changes in X could lead to large change in the parameters), and the variance of the predictors (beta hat) become large.
    
    * normality of errors for inference: Errors are normally distributed. This assumption is mainly needed for exact small-sample inference, such as t-tests and confidence intervans. If the assumption breaks, the statistics may not be exact.

    * Exogeneity/ Zero Conditional Mean: E[ϵ∣X]=0, after controlling for the features X, the remaining errors has no systematic relationship with x. Equivalently, the model is not leaving out some hidden factor that is both related to X and y. This makes sure beta is unbiased.

5. when assumptions matter for prediction vs inference \
  Linearity assumption matters for prediction, because if it breaks, the prediction is likely biased. For inference, all five assumptions matter, because each of them affects different aspects of confidence on the estimators.

6. why outliers hurt \
  Outliers hurt because they steer the linear regression in the direction of the outlier. An unusual y value can dominate the loss, and an unusual x value can strongly pull the fitted line. The worst case is an influence point with both unusual x and y, they substantially change the coefficient estimates and can cause bias.

7. how to diagnose misspecification
  Classic 4-panel diagnostic plot:
    * residual vs. fitted: y - y_hat vs. y_hat: This checks if the residuals roughly have a 0 expectation everywhere and roughly constant variance. If the model is under or over estimating in some regions, then there is bias.  If the variance (width between residuals of the same (or at least similar) y_hat) is not constant, then there is heteroskedacity.

    * Q-Q plot: Checks if all the residuals put together form a normal distribution. It is a rough global diagonostic.

    * Scale-location plot: Plots the sqrt of the absolute value of the standardized residual against y_hat. Also a test for heteroskedacity. A standardized residual is a residual divided by an estimate of its standard deviation, so the plot puts residuals on a comparable scale.

    * Residual vs. Leverage plot: Leverage is calculated as the diagonal of the hat matrix. y_hat_i = summation_j (h_ij y_j), so the large the h_ii is, point i's predicted value depends more on its own y (training response value). This happens precisely when x is unusual. Residual vs. Leverage checks if the point is both far from the rest of the point, and the residual is unusually large (meaning it is not in sync with the rest of the point).

8. interaction terms
  The interaction term lets the effect of 1 feature on the response variable to be dependent on the other feature.

9. polynomial features
  Polynomial features allow linear models to capture non-linear relationships between the response and the independent variable while staying linear in the coefficients.

10. dummy variables
  For category variables with k different possible values to be treated as numerical, they are converted to k-1 (usually binary) dummy or indicator columns. You learn a coeff for each dummy variable, and in inference, depending on which categories the input has, only those coeffs will take effect. Also note that we use k-1, not k, to avoid perfect collinearity between those variables.

11. feature scaling effects
  Scaling does not affect predictions or R^2, but it does affect regularization or convergence rate (if using GD like optimizer). Need to apply scaling before those.

### 2.2 Logistic regression
Logistic regression is a classification model. The core assumption for logistic regression is that we belive different things (*features*) affect **odds** in a **multiplicative** way. For example, if an event A makes the odds 2 times larger, and event B makes the odds 3 times larger, then we assume they together will make the odds 2 * 3 = 6 times larger.


1. sigmoid \
$$\sigma(z) = \frac{1}{1 + e^{-z}}$$
squashes any real number into (0, 1)

| Property | Value |
|----------|-------|
| $\sigma(0)$ | $0.5$ |
| $\sigma(\infty)$ | $1$ |
| $\sigma(-\infty)$ | $0$ |
| $\sigma(-z)$ | $1 - \sigma(z)$ (symmetry) |
| $\sigma'(z)$ | $\sigma(z)(1 - \sigma(z))$ (elegant!) |

2. log-odds \
Assuming $$p = P(y = 1)$$

define $\text{odds} = \frac{p}{1-p}$

It 1) has some interpretable meaning, i.e. how likely is class 1 relative to class 0. and 2) If we just model p, it's only going to take values between 0 and 1, and if we model p/(1-p), it takes values between 0 to inf.

Now, we mentioned about logistics regression assumes features operate on odds in a multipliative way, by taking log so we can turn right hand side into a **linear model**. (i.e. log(A * B * C) = log(A) + log(B) + log(C), which we can model with $\beta_0 + \beta_1 x_1 + \beta_2 x_2$.

Formally,
$$\underbrace{\log\frac{P(y=1)}{P(y=0)}}_{\text{log-odds (logit)}} = \mathbf{w}^\top \mathbf{x} + b$$

A mathematical side note is that taking log also makes the range of values odds can take from (0, inf) to (-inf, inf), which fully matches the range for linear models.

Finally, we can solve back p,
$$p = \frac{e^{\mathbf{w}^\top \mathbf{x}+b}}{1 + e^{\mathbf{w}^\top \mathbf{x}+b}} = \sigma(\mathbf{w}^\top \mathbf{x}+b)$$

So we are effectively shrinking the linear model to between 0 and 1 with the sigmoid function.

3. decision boundary \
What is decision bounday: It's a line curve, or surface where the model is indifferentiable between the two classes.

For logistic regression, it depends on the threshold you choose. When the threshold is the usual 0.5, the boundary is the contour $\mathbf{w}^\top \mathbf{x}+b = 0$. This is precisely when P(y = 0) = p(y = 1) = 0.5.

Note: Changing the threshold changes the decision boundary in feature space, but it does not change the model's underlying probability surface.

4. cross-entropy / log loss
We model each label $y_i \in \{0,1\}$ as a Bernoulli with probability $\hat{p}_i = \sigma(\mathbf{w}^\top \mathbf{x}_i + b)$.

The likelihood of the observing the data is $$\mathcal{L}(\mathbf{w}) = \prod_{i=1}^{n} \hat{p}_i^{y_i}(1-\hat{p}_i)^{1-y_i}$$

Taking the log (easier to optimize, same argmax):
$$\log \mathcal{L}(\mathbf{w}) = \sum_{i=1}^{n} \left[ y_i \log \hat{p}_i + (1-y_i)\log(1-\hat{p}_i) \right]$$

Finally, we usually want to minimize the loss, so maximizing the log likelihood is the same as minimizing the negative log likelihood. Taking the avergae, we get

$$\mathcal{J}(\mathbf{w}) = -\frac{1}{n}\sum_{i=1}^{n} \left[ y_i \log \hat{p}_i + (1-y_i)\log(1-\hat{p}_i) \right]$$

which is the cross-entropy loss.

5. why it is linear in feature space \
Because the log odds and the features have a linear relationship. Logistic regression has a linear decision boundary in the feature space.

7. thresholding vs probability estimation

Probability Estimation are given by the model, however classification is given by the thresholding after the probabilities are calculated. So you can have a model that outputs p = 0.69 for some example, but if you pick p >= 0.7 as the threshold for positive, then this example is still "negative".

Now, the model is still trained with cross entropy loss to learn to output probabilities. Howeverm, a model can do better in CE loss and worse in thresholded classification. For example, Model **A** gives high probability to all examples (e.g. > 0.8 for positives, < 0.2 for negatives), but makes 1 mistake (e.g. assigns 0.9 to a negative example). Model **B** makes mild but correct probabilities to all examples (e.g. 0.6 to all positive examples, and 0.4 for all negative examples). With proper thresholding model B does better than model A in practice if the objective is actually *to correctly classify examples*.

In practice, you have a separate validation set to evaluate your model's true performance on realistic metrics (e.g. if the goal is instead to come up with risk scores or ranking of all examples, then the raw probabilities matter). If the task involves thresholding, you can use the proper augmatation graphs (e.g. ROC curve, precision/recall curve) to come up with the best threshold.

8. class imbalance behavior \
When you have class imbalance, then assigning everything to the majority class is going to give a low loss - and the model might actually learn to do that.

To avoid it, one can
* Upweight minority class - give higher weights in the loss function.
* Oversample minority or undersample majority.
* Adjust the decision boundary to flag more minority class. (In tern force the model to learn the adjusted decision boundary).

9. calibration properties
Calibration asks: among all examples where the model predicts p_hat = 0.x, exactly x% should be positive.

If yes, the model is well-calibrated at 0.x.

## 4. Tree-based models

### 4.1 Decision trees
The training objective of decision trees is to maximize reduction in node impurity with each split.  

Information gain (impurity reduction) is calculated as:
$$
\text{IG}(I) = I(\mathcal{D}_t) - \left[\frac{|\mathcal{D}_L|}{|\mathcal{D}_t|} I(\mathcal{D}_L) + \frac{|\mathcal{D}_R|}{|\mathcal{D}_t|} I(\mathcal{D}_R)\right]
$$
where *I* is typically one of 1. gini impurity, 2. entropy, or 3. Mean squared error (for regression).

1. greedy splitting \
It means to split based on the current best option without thinking far ahead. A suboptimal split at current level might lead to optimal split overall, but that is NP-hard. Greedy splitting makes the problem tractable but might miss globally better solutions.

2. impurity measures:
  * gini impurity:
  $$
G = 1 - \sum_{k=1}^{K} p_k^2
$$

where $p_k$ is the fraction of samples in class $k$, and $K$ is the number of classes. It measure how "scattered" classes are.
  * entropy
  Entropy measures the average number of bits needed to encode the class label of a randomly drawn sample from that node, assuming an optimal coding scheme.
$$
H = -\sum_{k=1}^{K} p_k \log_2 p_k
$$

(Convention: $0 \cdot \log 0 = 0$.)


3. regression trees and variance reduction \
The impurity function is the **mean squared error** within a node:

$$
\text{MSE}(\mathcal{D}_t) = \frac{1}{|\mathcal{D}_t|} \sum_{i \in \mathcal{D}_t} (y_i - \bar{y}_t)^2
$$

where $\bar{y}_t = \frac{1}{|\mathcal{D}_t|}\sum_{i \in \mathcal{D}_t} y_i$ is the node mean.

The prediction at each leaf is usually the mean target value of the training samples in that leaf.


4. stopping criteria \
* Max depth: When the tree reaches a certain depth.
* Min samples split: Only split a node if it has at least min_sample_split samples.
* Min samples leaf: Each leaf node must have at least min_samples_leaf samples.
* min impurity decrease: Split only if $\Delta I \geq \epsilon$
* Max leaf number: Total number of leaves must be less than this number.

5. pruning \
Instead of early stopping, we can let the tree fully grow and prune the tree afterwards:

Define the **cost-complexity criterion** for a tree $T$:

$$
R_\alpha(T) = R(T) + \alpha \cdot |T|
$$

We train the tree with different alpha and pick the best alpha with cross validation.


6. why trees overfit \
Trees are highly flexible nonparametric learners. They are non-linear and can learn any shaped data with splits. So, if there is no complexity constraints, they can simply memorize the data and overfit. They also have high-variance, meaning small changes in training data can largely alter the shape of the tree. That is why ensemble methods like gradient boosting or random forests are used to reduce variance. Regularization methods also help.

7. handling nonlinearity and interactions \
A tree discovers interactions because later splits depend on earlier splits. Trees capture nonlinear effects because they capture feature space into regions.

8. why trees do not need feature scaling \
The infomration gain depends on the split which is independent of the feature scale.

4.2  Random Forests
1. bagging \
bagging = bootstrap aggregating
* Bootstrap many different training sets from the original data.
* Train one model on each sampled dataset.
* Average their predictions for regression, or majority vote for classification.

2. bootstrap sampling
Sample with replacement and result in the same sized data as the original dataset. Each sample contains roughly 63.2% unique observations from the original dataset. The remaining roughly 36.8% are out-of-bag for that tree (usually for validation).

3. feature subsampling
Random forest adds another layer of randomness. At each split (NOT once per tree, this is another method), you randomly sample a subset of features and only pick from those features the best split that reduces impurity scores. This further reduces variance.

4. variance reduction \
Variance reduction comes from 2 parts:
* Decorrelate between trees.
* Averaging between many trees.

Assuming each tree has the same variance $\sigma^2$. Here variance means if you construct different trees from different sampled data and features for splits, how spread will the prediction be.

$$
\operatorname{Var}\left(\frac{1}{B}\sum_{b=1}^{B} h_b(x)\right)
= \rho \sigma^2 + \frac{1 - \rho}{B}\sigma^2
$$

The former is controlled by correlation between trees, the latter shrinks as number of trees increase.


5. out-of-bag estimate \
OOB validation works like this: for each data that is not used in training in each tree, we treat as validation set, and use the trained tree to predict a value. Do this for all trees and for all untrained data. Then for points that is in the validation set for multiple trees, we still use aggregation (avging or voting) to get a single prediction. Then we aggregate prediction for all points to get a validation score. We can use the validation score to tune the hyperparameters for the whole forest.

6. why random forests are more stable than single trees \
Because they are trained on purturbed datasets and average the predictions. It uses many plausible versions of the training set and many feature subsets and aggregates results from all of them, which makes predictions smoother.

7. feature importance caveats \
(Fundamentally a decision tree thing, but more saliant in random forest)
Continuous features with many different ways to split naturally get more 'tries' in a split, so they might have a higher chance of 'reducing impurity' and look important. Categorical features with few different values on the other hand have less splits to offer. This might create interpretability bias (mistakenly perceiving some features as more important than others). Can use permutation important to help solve address the problem (randomly switching values of a feature in the data and see how badly it affects model's prediction).


## Gradient boosting
1. boosting intuition \
Boosting combines all previous trees, but usually as a weighted sum of corrections. Each tree is trained sequentially, paying attention to what the eariler trees got wrong.

2. sequential error correction \
Gradient boosting works by first making a (weak) generic prediction, then gradually corrects the model in the direction of the negative gradient of the loss function. If the loss is MSE, then the negative gradient is just the residual.


3. gradient boosting vs bagging \
Bagging builds many independent trees in parallel and aggregate their results, boosting build trees sequentially and each depends on the previous tree. Later trees add weighted correction to the mistake made by earlier trees.

4. learning rate/shrinkage & number of trees \
Those two are usually tuned together. Start by picking a set of learning rates, then as you build with more trees, watch validation loss. Pick number of trees for each learning rate with validation loss stops decreasing. Then choose the best learning rate /number of trees pair as the collective best set of hyperparameters.

5. tree depth \
Boosting usually uses shallow trees because each tree is meant to be a small correction. If we use deep trees, they might overfit and fit pseudo-gradients too aggressively.

6. overfitting behavior \
Overfitting happens when training loss keeps improving but not validation loss. Can be solved by tuning hyperparameters, or regularizing on number of leaves, or leaf size.
